```python
alpha_w = widgets.Dropdown(options=sorted(df_total['Alpha'].unique()), description='Alpha:')
var_w = widgets.Dropdown(options=['net_toa', 'swcf', 'lwcf', 'rsnt', 'rlnt'], description='Variabile:')

def update_dashboard_comparison(alpha, var):
    alpha_tag = str(float(alpha)).replace('.', '')
    alpha_folder = f"{alpha:.2f}".replace('.', '')
    
    metrics = ['l1', 'l2']
    
    display(HTML(f"<h2>METRICS COMPARISON | Alpha: {alpha} | Var: {var}</h2>"))
    
    summary_data = df_total[(df_total['Alpha'] == alpha)].copy()
    
    for m in metrics:
        img_name = f"comparison_2d_{var}_Global_a{alpha_tag}.png"
        img_path = os.path.join(results_dir, m, "net_toa_tol", 
                                f"alfa_{alpha_folder}", img_name)
        
        display(HTML(f"<div style='background-color: #f0f0f0; padding: 10px; margin-top: 20px; border-left: 5px solid #333;'>"
                     f"<h3>Metrics: {m.upper()}</h3></div>"))
    
        row = summary_data[summary_data['Metric'].str.lower() == m.lower()]
        if not row.empty:
            display(row[[f'{var}_AbsBias', f'{var}_PhysSpatial', 'Total_Objective_Score', 'Metric_Type']])
        
        if os.path.exists(img_path):
            display(Image(filename=img_path, width=1100))
        else:
            print(f"❌ Maps not found: {img_path}")

widgets.interact(update_dashboard_comparison, alpha=alpha_w, var=var_w);
```

## PROVA STAMPA MAPPA BIAS REALIZED VS PREDICTED

```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import os
import glob
import sys

path_to_utils = '/ec/res4/hpcperm/ecme3038/ecearth/ecearth4/ECtuner/ectuner/utils/'
if path_to_utils not in sys.path:
    sys.path.append(path_to_utils)

from sensitivity_2D import regrid_to_regular_smm_safe

def plot_emulator_residual_map(tuner_diag_file, scratch_dir, exp_tag, year1, year2, ceres_file, var_tuner, output_dir="./"):
    """
    2d residual map generation using the new monthly CERES EBAF dataset.
    Temporal slices aligned between model and observations.
    """
    # 1. predicted bias
    ds_tuner = xr.open_dataset(tuner_diag_file)
    predicted_bias = ds_tuner[f'{var_tuner}_bias_final'] 

    # 2. real flux computing from scratch
    pattern = os.path.join(scratch_dir, f"{exp_tag}_atm_cmip6_1m_*.nc")
    files = sorted(glob.glob(pattern))
    valid_files = [f for f in files if year1 <= int(os.path.basename(f).split('_')[-1].split('-')[0]) <= year2]
    
    print(f"--- Aggregating {len(valid_files)} years from scratch for {exp_tag} ---")
    ds_model = xr.open_mfdataset(valid_files, combine='nested', concat_dim='time_counter', chunks={'time_counter': 12})
    if 'time_counter' in ds_model.dims: 
        ds_model = ds_model.rename({'time_counter': 'time'})

    if var_tuner == 'net_toa':
        model_flux = (ds_model['rsdt'] - ds_model['rsut'] - ds_model['rlut']).mean('time').compute()
    elif var_tuner == 'rsnt':
        model_flux = (ds_model['rsdt'] - ds_model['rsut']).mean('time').compute()
    elif var_tuner == 'rlnt':
        model_flux = (-1.0 * ds_model['rlut']).mean('time').compute()
    elif var_tuner == 'swcf':
        model_flux = (ds_model['rsutcs'] - ds_model['rsut']).mean('time').compute()
    elif var_tuner == 'lwcf':
        model_flux = (ds_model['rlutcs'] - ds_model['rlut']).mean('time').compute()

    # 3. regridding
    print("--- Regridding model flux via smmregrid ---")
    model_reg = regrid_to_regular_smm_safe(model_flux, 'r180x90', method='ycon', 
                                           raw_file_path=valid_files[0], varname=var_tuner)

    # 4. ceres climatology loading and processing
    print(f"--- Processing New CERES Dataset for years {year1}-{year2} ---")
    ds_ceres = xr.open_dataset(ceres_file)
    
    # ceres preprocessing
    ceres_sliced = ds_ceres.sel(time=slice(f"{year1}-01-01", f"{year2}-12-31"))
    ceres_clim = ceres_sliced.mean('time')
    
    # spatial alignment
    if model_reg.lon.min() < 0 and ceres_clim.lon.max() > 180:
        ceres_clim = ceres_clim.assign_coords(lon=(((ceres_clim.lon + 180) % 360) - 180)).sortby('lon')
    
    ceres_clim = ceres_clim.interp(lat=model_reg.lat, lon=model_reg.lon, method='linear')

    if var_tuner == 'net_toa':
        ceres_flux = ceres_clim['toa_net_all_mon']
    elif var_tuner == 'rsnt':
        ceres_flux = ceres_clim['solar_mon'] - ceres_clim['toa_sw_all_mon']
    elif var_tuner == 'rlnt':
        ceres_flux = -1.0 * ceres_clim['toa_lw_all_mon']
    elif var_tuner == 'swcf':
        ceres_flux = -1.0 * ceres_clim['toa_cre_sw_mon']
    elif var_tuner == 'lwcf':
        ceres_flux = -1.0 * ceres_clim['toa_cre_lw_mon']

    # real bias computing and alignment
    realized_bias = model_reg - ceres_flux
    predicted_bias_aligned = predicted_bias.reindex_like(realized_bias, method='nearest')
    residual_map = realized_bias - predicted_bias_aligned

    fig = plt.figure(figsize=(12, 7))
    ax = fig.add_subplot(1, 1, 1, projection=ccrs.PlateCarree())
    
    vmax = float(np.abs(residual_map).max()) * 0.8  
    vmin = -vmax
    
    im = residual_map.plot(ax=ax, transform=ccrs.PlateCarree(), cmap='bwr', 
                           vmin=vmin, vmax=vmax, add_colorbar=False)
    
    ax.coastlines(resolution='110m', color='black', linewidth=1)
    ax.gridlines(draw_labels=True, linestyle=':', alpha=0.5)
    
    cbar = fig.colorbar(im, ax=ax, orientation='horizontal', pad=0.08, fraction=0.05)
    cbar.set_label(f'Emulator Residual (Realized Bias - Predicted Bias) [W/m²]', fontsize=11)
    
    plt.title(f'2D Spatial Residual of the Linear Emulator)\nVariable: {var_tuner.upper()} | Run: {exp_tag} ($\\alpha=0.6$)', 
              fontsize=13, fontweight='bold')
    
    output_path = os.path.join(output_dir, f"emulator_residual_synchronized_{var_tuner}_{exp_tag}.png")
    plt.savefig(output_path, bbox_inches='tight', dpi=200)
    print(f"--- Mappa salvata con successo: {output_path} ---")
    plt.show()
```

## PROVA RSDT

```python
import xarray as xr
import glob

# Percorsi ai raw files (sostituisci YOUR_PERT_EXP con il nome di un run perturbato, es. nw00)
path_ref = "/ec/res4/scratch/ecme3038/ece4/phis/output/oifs/phis_atm_cmip6_1m_2000*.nc"
path_pert = "/ec/res4/scratch/ecme3038/ece4/nw09/output/oifs/nw09_atm_cmip6_1m_2000*.nc"

# Caricamento del run di riferimento
files_ref = sorted(glob.glob(path_ref))
ds_ref = xr.open_mfdataset(files_ref, use_cftime=True)
rsdt_ref = ds_ref['rsdt'].mean().compute()

# Caricamento del run perturbato
files_pert = sorted(glob.glob(path_pert))
ds_pert = xr.open_mfdataset(files_pert, use_cftime=True)
rsdt_pert = ds_pert['rsdt'].mean().compute()

print(f"RSDT Baseline (phis): {float(rsdt_ref):.4f} W/m2")
print(f"RSDT Perturbato     : {float(rsdt_pert):.4f} W/m2")
```

```python
import xarray as xr
import numpy as np

# Inserisci il nome esatto del file, includendo lo spazio
file_path = '/ec/res4/hpcperm/ecme3038/ecearth/ecearth4/ECtuner/ectuner/ecmean/reference/climatology/EC26-PDAY/r180x90/climate_average_rlnt_CERES EBAF_r180x90_2000-2024.nc'

# 1. Carica il dataset
ds = xr.open_dataset(file_path)

# 2. Estrai la variabile (assicurati che internamente si chiami 'rlnt')
# Se c'è una dimensione temporale, facciamo prima la media nel tempo
if 'time' in ds.dims:
    var = ds['rlnt'].mean('time')
else:
    var = ds['rlnt']

# 3. Calcola i pesi spaziali (coseno della latitudine)
weights = np.cos(np.deg2rad(ds.lat))
weights.name = "weights"

# 4. Calcola la media globale pesata e stampa
global_mean = var.weighted(weights).mean().values

print(f"Media globale pesata per {file_path}: {global_mean:.4f} W/m²")
```
