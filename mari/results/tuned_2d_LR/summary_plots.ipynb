```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import ipywidgets as widgets
from IPython.display import display, Image, HTML
import os

#functions
from compare_tuning import extract_alpha, load_results
```

```python
results_dir = ""
config_ref = "../../ectuner/myconfigs/myconfig_415_2d_LR.yaml"
dir_l1 = os.path.join(results_dir, "l1/net_toa_tol/yaml_files")
dir_l2 = os.path.join(results_dir, "l2/net_toa_tol/yaml_files")
```

```python
#  L1
df_l1, params = load_results(dir_l1, config_ref)
if df_l1 is not None: df_l1['Metric_Type'] = 'L1 (Linear)'

# L2
df_l2, _ = load_results(dir_l2, config_ref)
if df_l2 is not None: df_l2['Metric_Type'] = 'L2 (Quadratic)'

df_total = pd.concat([df_l1, df_l2]).sort_values(by=['Metric_Type', 'Alpha'])

df_plot = df_total[df_total.index != 'REFERENCE_OIFS']

print("data loaded!")
pd.set_option('display.max_columns', None)
display(df_total)
```

```python
def plot_global_comparison():
    plt.figure(figsize=(11, 7))
    
    for m_type in df_plot['Metric_Type'].unique():
        subset = df_plot[df_plot['Metric_Type'] == m_type]
        
        # Usiamo le colonne 'Phys' che abbiamo creato in load_results
        plt.plot(subset['Phys_Spatial_Total'], subset['Phys_Global_Total'], 
                 marker='o', linestyle='--', label=m_type)
        
        # Annotazioni Alpha
        for _, row in subset.iterrows():
            plt.annotate(f"a={row['Alpha']:.2f}", 
                         (row['Phys_Spatial_Total'], row['Phys_Global_Total']),
                         textcoords="offset points", xytext=(0,10), ha='center', fontsize=8)

    plt.title("Confronto Fronti di Pareto: L1 vs L2 (Unità Fisiche)")
    plt.xlabel("Weighted Spatial Error (RMSE/MAE) [$W/m^2$]")
    plt.ylabel("Weighted Global Bias [$W/m^2$]")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()

plot_global_comparison()
```

```python
# Widget per la selezione
metric_w = widgets.Dropdown(options=['l1', 'l2'], description='Metrica:')
alpha_w = widgets.Dropdown(options=sorted(df_plot['Alpha'].unique()), description='Alpha:')
var_w = widgets.Dropdown(options=['net_toa', 'swcf', 'lwcf', 'rsnt', 'rlnt'], description='Variabile:')

def update_dashboard(metric, alpha, var):
    alpha_str = f"{alpha:.2f}".replace('.', '')
    
    # MODIFICA: Il percorso ora deve riflettere la nuova gerarchia
    # results_dir dovrebbe essere "../results/tuned_2d_LR/"
    img_path = os.path.join(results_dir, metric.upper(), "net_toa_tol", 
                            f"alfa_{alpha_str}", f"comparison_2d_{var}_global_a{alpha_str}.png")
    
    display(HTML(f"<h3>Esperimento: {metric.upper()} | Alpha: {alpha} | Var: {var}</h3>"))
    
    # Mostra i dati numerici specifici
    row = df_total[(df_total['Metric'] == metric) & (df_total['Alpha'] == alpha)]
    if not row.empty:
        display(row[[f'{var}_AbsBias', f'{var}_PhysSpatial', 'Metric_Type']])
    
    if os.path.exists(img_path):
        display(Image(filename=img_path, width=1000))
    else:
        print(f"Mappa non trovata in: {img_path}")

widgets.interact(update_dashboard, metric=metric_w, alpha=alpha_w, var=var_w);
```
