```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import ipywidgets as widgets
from IPython.display import display, Image, HTML
import os

#functions
import compare_tuning as ct
```

```python
results_dir = ""
config_ref = "../../ectuner/myconfigs/myconfig_415_2d_LR.yaml"
dir_l1 = os.path.join(results_dir, "l1/all_flux_tol/yaml_files")
dir_l2 = os.path.join(results_dir, "l2/all_flux_tol/yaml_files")
```

```python
#  L1
df_l1, params = ct.load_results(dir_l1, config_ref)
if df_l1 is not None: df_l1['Metric_Type'] = 'L1 (Linear)'

# L2
df_l2, _ = ct.load_results(dir_l2, config_ref)
if df_l2 is not None: df_l2['Metric_Type'] = 'L2 (Quadratic)'

df_total = pd.concat([df_l1, df_l2]).sort_values(by=['Metric_Type', 'Alpha'])

df_plot = df_total[df_total.index != 'REFERENCE_OIFS']

print("data loaded!")
pd.set_option('display.max_columns', None)
#display(df_total)
display(df_l1)
```

```python
display(df_l2)
```

```python
# Widget per la selezione
alpha_w = widgets.Dropdown(options=sorted(df_plot['Alpha'].unique()), description='Alpha:')
var_w = widgets.Dropdown(options=['net_toa', 'swcf', 'lwcf', 'rsnt', 'rlnt'], description='Variabile:')

def update_dashboard_comparison(alpha, var):
    # Tag per i file e le cartelle (logica robusta)
    alpha_tag = str(float(alpha)).replace('.', '')
    alpha_folder = f"{alpha:.2f}".replace('.', '')
    
    metrics = ['l1', 'l2']
    
    display(HTML(f"<h2>METRICS COMPARISON | Alpha: {alpha} | Var: {var}</h2>"))
    
    # Creiamo due colonne per i dati numerici (opzionale ma pulito)
    summary_data = df_total[(df_total['Alpha'] == alpha)].copy()
    
    for m in metrics:
        # Costruzione percorsi
        img_name = f"comparison_2d_{var}_Global_a{alpha_tag}.png"
        # Usiamo .upper() o .lower() a seconda di come hai chiamato fisicamente le cartelle
        img_path = os.path.join(results_dir, m, "net_toa_tol", 
                                f"alfa_{alpha_folder}", img_name)
        
        display(HTML(f"<div style='background-color: #f0f0f0; padding: 10px; margin-top: 20px; border-left: 5px solid #333;'>"
                     f"<h3>Metrics: {m.upper()}</h3></div>"))
        
        # 1. Mostra dati numerici
        row = summary_data[summary_data['Metric'].str.lower() == m.lower()]
        if not row.empty:
            display(row[[f'{var}_AbsBias', f'{var}_PhysSpatial', 'Total_Objective_Score', 'Metric_Type']])
        
        # 2. Mostra l'immagine
        if os.path.exists(img_path):
            display(Image(filename=img_path, width=1100))
        else:
            print(f"❌ Maps not found: {img_path}")
            # Debug per capire cosa non va nel path
            # print(f"DEBUG: cercavo in {os.path.abspath(img_path)}")

# Lanciamo l'interazione senza metric_w
widgets.interact(update_dashboard_comparison, alpha=alpha_w, var=var_w);
```

```python
# Widget per il confronto tra due Alpha differenti
metric_w = widgets.Dropdown(options=['l1', 'l2'], description='Metrica:')
alpha1_w = widgets.Dropdown(options=sorted(df_total['Alpha'].unique()), description='Alpha A:', value=0.0)
alpha2_w = widgets.Dropdown(options=sorted(df_total['Alpha'].unique()), description='Alpha B:', value=1.0)
var_w = widgets.Dropdown(options=['net_toa', 'swcf', 'lwcf', 'rsnt', 'rlnt'], description='Variabile:')

def update_alpha_comparison(metric, a1, a2, var):
    display(HTML(f"<h2>ALPHA COMPARISON ({metric.upper()}) | {var}</h2>"))
    
    for alpha in [a1, a2]:
        alpha_tag = str(float(alpha)).replace('.', '')
        alpha_folder = f"{alpha:.2f}".replace('.', '')
        img_name = f"comparison_2d_{var}_Global_a{alpha_tag}.png"
        img_path = os.path.join(results_dir, metric, "net_toa_tol", f"alfa_{alpha_folder}", img_name)
        
        display(HTML(f"<div style='background-color: #e8f4f8; padding: 5px; margin-top: 15px; border-left: 5px solid #2980b9;'>"
                     f"<h3>Alpha: {alpha}</h3></div>"))
        
        # Dati numerici
        row = df_total[(df_total['Metric'].str.lower() == metric.lower()) & (df_total['Alpha'] == alpha)]
        if not row.empty:
            display(row[[f'{var}_AbsBias', f'{var}_PhysSpatial', 'Total_Objective_Score']])
        
        # Immagine
        if os.path.exists(img_path):
            display(Image(filename=img_path, width=1000))
        else:
            print(f"❌ Mappa non trovata: {img_path}")

widgets.interact(update_alpha_comparison, metric=metric_w, a1=alpha1_w, a2=alpha2_w, var=var_w);
```
